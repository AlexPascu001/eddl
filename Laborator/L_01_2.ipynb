## L_01_2. The ```cities``` dataset

The aim of the notebook is to illustrate the ```cities``` dataset, available in the geomstats repository. It is related to the sphere, which is a surface (a 2-dimensional manifold).

````python
### Initialization 
import os
import subprocess
print('Working directory: ', os.getcwd())
import geomstats.backend as gs
import matplotlib
import matplotlib.image as mpimg
import matplotlib.pyplot as plt
import geomstats.visualization as visualization
````

````python
### Load the cities dataset
import geomstats.datasets.utils as data_utils
data, names = data_utils.load_cities()
print(names[:10])
print(data[:10])
````

````python
### The data belongs to the sphere S2
from geomstats.geometry.hypersphere import Hypersphere
sphere = Hypersphere(dim=2)
gs.all(sphere.belongs(data))
````

````python
### Visualization
fig = plt.figure(figsize=(20, 15))
ax = fig.add_subplot(111, projection='3d')
visualization.plot(data, ax=ax, space='S2', s=80, alpha=0.5)
visualization.plot(data[15:20], ax=ax, space='S2', label=names[15:20], s=80, alpha=0.5)
ax.set_title('Cities on the earth.');
````

````python
### Exploration: cities dataset
import geomstats.datasets.utils as data_utils
from geomstats.geometry.hypersphere import Hypersphere

cities_data, cities_names = data_utils.load_cities()
print("="*60)
print("CITIES DATASET ANALYSIS")
print("="*60)
print(f"Number of cities in the dataset: {len(cities_data)}")
print(f"Shape of data: {cities_data.shape}")
print("\nFirst 10 cities:")
for i in range(min(10, len(cities_names))):
    print(f"  {i + 1}. {cities_names[i]}")

sphere = Hypersphere(dim=2)
print(f"\nData points belong to S^2: {gs.all(sphere.belongs(cities_data))}")
print("\nSample coordinates (first 5 cities):")
print(cities_data[:5])
````
