## L_01_1. Introduction

The aim of this notebook is to handle some basic manifolds (plane R^2, space R^3, sphere) and to apply basic operations such as the mean. 

### Initializations

```python
### Initialization 
import os
import subprocess
print('Working directory: ', os.getcwd())
import geomstats.backend as gs
import geomstats.visualization as visualization
visualization.tutorial_matplotlib()

import matplotlib
import matplotlib.image as mpimg
import matplotlib.pyplot as plt
from matplotlib import rcParams

# Global rcParams
rcParams["font.family"] = "sans-serif"
rcParams["font.serif"] = ["Times New Roman"]
rcParams["font.sans-serif"] = ["Arial"]
rcParams["font.monospace"] = ["Courier New"]
rcParams["mathtext.fontset"] = "custom"
rcParams["mathtext.rm"] = "Times New Roman"
rcParams["mathtext.it"] = "Times New Roman:italic"
rcParams["mathtext.bf"] = "Times New Roman:bold"

gs.random.seed(2020)
```

### Example 1. The Euclidean plane R^2

```python
### EXAMPLE 1:
### R2=Euclidean space - 2D
from geomstats.geometry.euclidean import Euclidean
dim = 2
n_samples = 100
manifold = Euclidean(dim=dim)
```

```python
### Select randomly sampled points
points_on_manifold = manifold.random_point(n_samples=n_samples)
print('Points on the manifold:\n', points_on_manifold)
```

```python
### Linear mean on the manifold
linear_mean = gs.sum(points_on_manifold, axis=0) / n_samples
print(manifold.belongs(linear_mean))
print('Mean of points:\n', linear_mean)
```

```python
### Graphic representation

%matplotlib inline
fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(111)
ax.scatter(points_on_manifold[:, 0], points_on_manifold[:, 1], label='Points')
ax.scatter(
    gs.to_numpy(linear_mean[0]),
    gs.to_numpy(linear_mean[1]), label='Mean', s=80, alpha=0.5)
ax.set_title('Mean of points in E2')
ax.legend();
```

### Example 2. The Euclidean space R^3

```python
### EXAMPLE 2:
### R3=Euclidean space - 3D
from geomstats.geometry.euclidean import Euclidean
dim = 3
n_samples = 200
manifold = Euclidean(dim=dim)
```

```python
### Select randomly sampled points
points_on_manifold = manifold.random_point(n_samples=n_samples)
print('Points on the manifold:\n', points_on_manifold)
```

### Example 3. The 2-dimensional sphere S^2

```python
### EXAMPLE 3:
### S2=2D-Sphere
from geomstats.geometry.hypersphere import Hypersphere
n_samples = 30
manifold = Hypersphere(dim=2)
```

```python
### Select randomly sampled points
points_on_manifold = manifold.random_point(n_samples=n_samples)
print('Points on the manifold:\n', points_on_manifold)
```

```python
### Linear mean on the manifold
linear_mean = gs.sum(points_on_manifold, axis=0) / n_samples
print(manifold.belongs(linear_mean))
print('Mean of points:\n', linear_mean)
```

```python
### Graphic representation
%matplotlib inline
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')
visualization.plot(points_on_manifold, ax=ax, space='S2', label='Point', s=80)
ax.scatter(points_on_manifold[:, 0], points_on_manifold[:, 1], points_on_manifold[:, 2], linestyle='dashed', alpha=0.5)
ax.scatter(linear_mean[0], linear_mean[1], linear_mean[2],label='Mean', s=80, alpha=0.5)
ax.set_title('Mean of points on the sphere S2')
ax.legend();
```
