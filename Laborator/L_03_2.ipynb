### Modification for the assignment
The original ENZYMES example was replaced with the MUTAG dataset.
MUTAG is a compact graph classification dataset built from molecular graphs, which makes it easy to inspect one sample graph and compare it with the original benchmark example.

```python
# Load dataset
from torch_geometric.datasets import TUDataset

dataset = TUDataset(root='data/TUDataset', name='MUTAG')
data = dataset[0]
```

```python
# Info about the dataset
print(dataset)
print(len(dataset))
print(f"Number of classes: {dataset.num_classes}")
print(f"Number of node features: {dataset.num_node_features}")
print(data)
print(f"Is data undirected? {data.is_undirected()}")
```
