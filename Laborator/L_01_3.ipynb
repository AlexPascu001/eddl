## L_01_3. The ```poses``` dataset

The aim of the notebook is to illustrate the ```poses``` dataset, available in the geomstats repository. This dataset is related to the manifold ${\rm SO}(3)$ - rotations in the 3D-space.

````python
### Initialization 
import os
import subprocess
print('Working directory: ', os.getcwd())
import geomstats.backend as gs
import matplotlib
import matplotlib.image as mpimg
import matplotlib.pyplot as plt
import geomstats.visualization as visualization
````

````python
### Load the dataset
import geomstats.datasets.utils as data_utils
data, img_paths = data_utils.load_poses()

img_path1, img_path2 = img_paths[4], img_paths[3]
img_path1 = os.path.join(data_utils.DATA_PATH, 'poses', img_path1)
img_path2 = os.path.join(data_utils.DATA_PATH, 'poses', img_path2)

img1 = mpimg.imread(img_path1)
img2 = mpimg.imread(img_path2)

fig = plt.figure(figsize=(16, 8))
ax = fig.add_subplot(121)
imgplot = ax.imshow(img1)
ax.axis('off')
ax = fig.add_subplot(122)
imgplot = ax.imshow(img2)
ax.axis('off')
plt.show()
````

````python
### The data belongs to the manifold SO(3)
from geomstats.geometry.special_orthogonal import SpecialOrthogonal
so3 = SpecialOrthogonal(n=3, point_type='vector')
gs.all(so3.belongs(data))
````

````python
### Plot orientations
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')
visualization.plot(data[:2], ax=ax, space='SO3_GROUP')
ax.set_title('3D orientations of the beds.');
````
